# Estructuras de Datos en Python

## LISTAS - Colecciones ordenadas y mutables

```python
print('='*50)
print('\t\t     LISTAS')
print('='*50)

# Creación de la lista llamada Ventas
ventas = [2000, 3500, 5000, 6500, 1800]

print(f'Ventas: {ventas}') # Imprime el contenido de la lista
print(f'Primera Venta: {ventas[0]}') # Imprime la venta que esta en índice 0
print(f'Última venta: {ventas[-1]}') # Imprime la venta que esta en el índice -1 (último)
print(f'Ventas del 2do al 4to mes: {ventas[1:4]}') # Imprime las ventas entre el rango 1:4
```

### Operaciones con Listas

```python
ventas.append(2800) # Agregar elemento
print(f'Después de agregar: {ventas}')

ventas.insert(0, 900) # Insertar elemento en una posición específica
print(f'Después de insertar: {ventas}')

ventas.remove(6500) # Eliminar un elemento
print(f'Después de eliminar: {ventas}')
```

### Otras Operaciones

```python
suma = sum(ventas)
promedio = suma / len(ventas)
maximo = max(ventas)
minimo = min(ventas)

print(f'Total de ventas: ${suma:,.2f}')
print(f'Promedio: ${promedio:,.2f}')
print(f'Máximo: ${maximo:,.2f}')
print(f'Mínimo: ${minimo:,.2f}')
```

## DICCIONARIOS - Colecciones de pares clave-valor

```python
print('='*50)
print('\t\t  DICCIONARIOS')
print('='*50)

# Creación de un diccionario
empleado = {
    "nombre": "Shalem Rincon",
    "edad": 18,
    "departamento": "Analytics",
    "salario": 5500,
    "habilidades": ['Python', 'SQL', 'Power BI']
}

print(f'Nombre empleado: {empleado["nombre"]}') # Imprimir nombre del empleado
print(f'Departamento: {empleado["departamento"]}') # Imprimir departamento del empleado
print(f'Habilidades: {empleado["habilidades"]}') # Imprimir las habilidades del empleado    
```

### Agregar una nueva clave

```python
empleado["años_experiencia"] = 5 # Se añade clave 'años_experiencia' con valor 5
print(f'Experiencia: {empleado['años_experiencia']} años')
```

### Iterar sobre diccionario

```python
print('Iteracción del diccionario empleado:')
for clave, valor in empleado.items():
    print(f'\t{clave}: {valor}')
```
